```python
# Cell 2 - clone repo in Colab
# You will be prompted for a GitHub PAT. The token is NOT saved to the notebook.
import os, sys, subprocess, getpass

REPO_HTTPS = "https://github.com/Sagar-Massachusetts/jnd-bo-exo-sim.git"
REPO_DIR   = "/content/jnd-bo-exo-sim"

if not os.path.isdir(REPO_DIR):
    token = getpass.getpass("GitHub PAT (repo scope): ")
    url = REPO_HTTPS.replace("https://", f"https://{token}@")
    subprocess.run(["git", "clone", url, REPO_DIR], check=True)
    # Remove token from remote URL after cloning so it's not stored
    subprocess.run(["git", "-C", REPO_DIR, "remote", "set-url", "origin", REPO_HTTPS], check=True)

os.chdir(REPO_DIR)
sys.path.insert(0, REPO_DIR)
print("Working dir:", os.getcwd())
```
